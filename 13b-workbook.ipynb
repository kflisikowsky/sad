# 📊 ZADANIE 2: Czy płeć wpływa na odpowiedzi?

## Pytanie badawcze:
**"Czy mężczyźni i kobiety różnią się w swoich opiniach na temat wybranego pytania?"**

## Zmienne:
- `plec` - płeć (1=mężczyzna, 2=kobieta)
- `q10` - dowolne pytanie opinii (wartości 1-4, gdzie 1=zdecydowanie tak, 4=zdecydowanie nie)

````python
# ZADANIE 2 - Twoje rozwiązanie tutaj

# Odfiltruj brakujące dane
df_clean = df[df['q10'].notna()].copy()

# Tabela kontyngencji dla zmiennej q10 i płci
cross_tab = pd.crosstab(df_clean['plec'], df_clean['q10'])

# Test statystyczny
result = stats.chi2_contingency(cross_tab)
print(result[0])  # Wartość statystyki chi-kwadrat
print(result[1])  # Wartość p
````

````python
# wykres z nalozonymi wynikami testu chi-kwadrat
fig, ax = plt.subplots(figsize=(8, 6))

cross_tab.rename(index={1: "Mężczyzna", 2: "Kobieta"}).plot.bar(
    ax=ax, colormap="Pastel1", rot=0
)

ax.set_title(
    f"Rozkład odpowiedzi q10 według płci\n"
    f"χ²({result[2]}) = {result[0]:.2f}, p = {result[1]:.3g}"
)
ax.set_xlabel("Płeć")
ax.set_ylabel("Liczba respondentów")
ax.legend(title="Czy obecnie pracuje Pan(i) zarobkowo?", loc="upper right")
fig.tight_layout()
plt.show()
````

# 📊 ZADANIE 3: Czy wiek wpływa na odpowiedzi?

## Pytanie badawcze:
**"Czy istnieje korelacja między wiekiem a odpowiedziami na pytania?"**

## Zmienne:
- `urodzony` - rok urodzenia respondenta
- `q12` - dowolne pytanie badania (skala ilościowa)

## 💡 Wskazówka

````python
# ZADANIE 3 - Twoje rozwiązanie tutaj

# Policzenie wieku
df['wiek'] = 2017 - df['urodzony']

# Korelacja
corr_result = stats.kendalltau(df['wiek'], df['q10'], nan_policy='omit')
print(f"Korelacja Kendalla tau: {corr_result.correlation:.3f}, p-value: {corr_result.pvalue:.3g}")
````

````python
# przekoduj zmienną q10 na kategorie 1= w pełnym wymiarze, 2= w niepełnym wymiarze, 3= dorywczo, 4= nie pracuję
df['q10'] = df['q10'].map({1: 'w pełnym wymiarze', 2: 'w niepełnym wymiarze', 3: 'dorywczo', 4: 'nie pracuję'})
````

````python
# boxplot 
sns.boxplot(hue='q10', y='wiek', data=df)
plt.title(f"Korelacja Kendalla tau: {corr_result.correlation:.3f}, p-value: {corr_result.pvalue:.3g}")
plt.xlabel("Czy obecnie pracuje Pan(i) zarobkowo?")
plt.ylabel("Wiek")  
plt.show()
````

## 📚 PRZYDATNE FUNKCJE I TESTY (rozdziały 1-12)

### Statystyka opisowa:
```python
df['zmienna'].describe()              # Statystyki podstawowe
df.groupby('grupa')['zmienna'].mean() # Średnia per grupa
df['zmienna'].value_counts()          # Rozkład (kategorie)
```

### Testy normalności (ROZDZIAŁ 12):
```python
pg.normality(x, method='shapiro')      # Test Shapiro-Wilka
pg.normality(x, method='normaltest')   # Test D'Agostino-Pearsona
```

### Testy porównujące grupy:
```python
pg.ttest(x, y)                        # T-test (parametryczny, 2 grupy)
pg.mannwhitneyu(x, y)                 # Mann-Whitney U (nieparametryczny, 2 grupy)
pg.wilcoxon(x, y)                     # Test Wilcoxona (dla prób powiązanych)
```

### Korelacja:
```python
pg.corr(x, y)                         # Korelacja Pearsona (parametryczna)
pg.corr(x, y, method='spearman')      # Korelacja Spearmana (nieparametryczna)
```

### Tabele kontyngencji:
```python
pd.crosstab(x, y)                     # Tabela krzyżowa
pg.chi2_independence(x, y)            # Test chi-kwadrat (niezależność)
```

### Wizualizacja (matplotlib + seaborn):
```python
sns.boxplot(data=df, x='grupa', y='zmienna')     # Box plot
sns.stripplot(data=df, x='grupa', y='zmienna')   # Punkty
sns.scatterplot(x='x', y='y', data=df)           # Scatter

```
